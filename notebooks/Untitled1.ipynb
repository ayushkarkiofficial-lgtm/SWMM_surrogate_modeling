```python
with open (r"D:\Claude\surrogate_modeling\swmm\base_model.rpt", "r") as file:
    lines = file.readlines()
    st, headings = [], []
    for n, line in enumerate(lines):
        if "***" in line:
            st.append(n)
    for id, i in enumerate(st):
        if id%2 == 0:
            headings.append(int((st[id]+st[id+1])/2))
    # print(headings)
    # [print(lines[i]) for i in headings]
    
def search_line(heading, item):
    for i, head in enumerate(headings):
        if heading.upper() in lines[head].upper():
            si = head
            ei = headings[i+1]
            continue
    for i in range(si+2, min(ei-1, len(lines))):
        # print(lines[i])
        if item.upper() in lines[i].upper():
            return lines[i].split()

peak_discharge_cfs = search_line("outfall loading", "out1")
total_flood_vol = search_line("Flow Routing Continuity", "Flooding Loss")
num_flooded_nodes = search_line("Node Flooding Summary", "out1")
continuity_error_pct = search_line("outfall loading", "out1")
print(peak_discharge_cfs, total_flood_vol, num_flooded_nodes, continuity_error_pct)
```
